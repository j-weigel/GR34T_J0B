```python
# Derivative f'(x) = -2x + 4
def df(x):
    return -2 * x + 4

# Bracket [a, b] containing root
a, b = 0.0, 3.0
tol = 1e-5

for step in range(20):
    c = (a + b) / 2.0
    f_c = df(c)
    if abs(f_c) < tol:
        break
    if df(a) * f_c > 0:
        a = c # Root is in [c, b]
    else:
        b = c # Root is in [a, c]
    print(f"Step {step}: c = {c:.4f}")
```
