## Multiple Peaks / Gradient Ascent :


```python
# Explore what happens when a surface has MORE THAN ONE peak.
#
# New surface:
#
#     f(x) = -(x + 2)^2 + 5 + 0.7 * sin(2x)
#
# The sinusoidal term creates bumps in the surface.
# This means the surface is no longer a simple parabola.
#
# For this session, we calculate the derivatives analytically.
def session_2_surface(x):
    return -(x + 2) ** 2 + 5 + 1.4 * np.sin(8 * x)


def session_2_df(x):
    return -2 * (x + 2) + 2.8 * np.cos(8 * x)


def session_2_ddf(x):
    return -2 - 5.6 * np.sin(8 * x)


# Define a general gradient ascent function that accepts the surface and derivative functions
def gradient_ascent(surface_func, derivative_func, start_x, learning_rate=0.1, steps=20):
    x = start_x
    history = [x]

    for i in range(steps):
        # Move x in the direction of the positive slope
        x = x + learning_rate * derivative_func(x)
        history.append(x)

    return x, history


def plot_surface(surface_func, x_min, x_max, title, path=None):
    x_vals = np.linspace(x_min, x_max, 400)
    y_vals = surface_func(x_vals)

    plt.figure(figsize=(10, 6))
    plt.plot(x_vals, y_vals, label='Surface')
    plt.title(title)
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid(True)

    if path is not None:
        path_x = np.array(path)
        path_y = surface_func(path_x)
        plt.plot(path_x, path_y, 'ro-', markersize=5, label='Gradient Ascent Path')
        plt.plot(path_x[0], path_y[0], 'go', markersize=8, label='Start')
        plt.plot(path_x[-1], path_y[-1], 'rx', markersize=10, label='End')
        plt.legend()
    plt.show()


# Run gradient ascent from THREE different starting points.
# This is intentional: starting location can matter on a complex surface.
session_2_starts = [-5, -2, 2]
session_2_results = []

print("\n" + "=" * 60)
print("PLAY SESSION 2: A BUMPY SURFACE")
print("=" * 60)

for start in session_2_starts:
    peak, path = gradient_ascent(
        session_2_surface,
        session_2_df,
        start_x=start,
        learning_rate=0.08,
        steps=80,
    )

    session_2_results.append((start, peak, path))

    print(
        f"Start x = {start:5.1f} -> "
        f"final x = {peak:8.4f}, "
        f"f(x) = {session_2_surface(peak):8.4f}"
    )

# Plot the three paths separately so we can compare how starting point
# changes the search.
for start, peak, path in session_2_results:
    plot_surface(
        session_2_surface,
        -6,
        3,
        f"Session 2: Gradient Ascent Starting at x={start}",
        path,
    )
```

## Gradient Descent:


```python
def df(x):
    return -2 * (x - 5) # Corrected derivative for f(x) = 10 - (x - 5)**2

def gradient_ascent(start_x, learning_rate=0.1, steps=20):
    x = start_x
    history = [x]

    for i in range(steps):
        # Move x in the direction of the positive slope
        x = x + learning_rate * df(x)
        history.append(x)
        if i % 5 == 0: # Print every 5 steps
            print(f"Step {i}: x = {x:.4f}")

    return x, history

# Try it out! You can change the learning_rate to see how it affects speed.
peak_grad, path = gradient_ascent(start_x=0.1, learning_rate=0.2)
print(f"\nGradient Ascent found peak at: {peak_grad:.4f}")
```

## Bisection Method:

```python
def bisection_method(a, b, tolerance=0.01):
    # We look for where the derivative df(x) is 0
    if df(a) * df(b) > 0:
        print("The peak might not be between these points!")
        return None

    while (b - a) / 2 > tolerance:
        midpoint = (a + b) / 2
        if df(midpoint) == 0:
            return midpoint
        elif df(a) * df(midpoint) < 0:
            b = midpoint # The peak is in the left half
        else:
            a = midpoint # The peak is in the right half

    return (a + b) / 2

# Try it out with a range from 0 to 5
peak_bisect = bisection_method(0, 5)
print(f"Bisection Method found peak near: {peak_bisect:.4f}")
```

## Newton's Method:


```python
def newtons_method(start_x, iterations=5):
    x = start_x
    print(f"Starting Newton's Method at x = {x}")

    for i in range(iterations):
        # Newton's update rule for optimization: x = x - f'(x) / f''(x)
        x = x - df(x) / ddf(x)
        print(f"Iteration {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}")

    return x

# Try it out!
peak_newton = newtons_method(start_x=0.5)
print(f"\nNewton found the peak at: {peak_newton} meters.")
```

Newton's Method was the fastest for this specific function.

Gradient Ascent is the most common in Machine Learning.

Bisection is the most reliable if you don't know the derivative.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

# ---------------------------------------------------------
# Define the terrain function
# ---------------------------------------------------------
def f(coords):
    x, y = coords
    return (x**2 - 2)**2 - x*y + (y**2 - 3)**2

# ---------------------------------------------------------
# Custom Nelder-Mead that logs every simplex (triangle) step
# ---------------------------------------------------------
def nelder_mead(f, x0, step=1.0, tol=1e-6, max_iter=60,
                 alpha=1.0, gamma=2.0, rho=0.5, sigma=0.5):
    dim = len(x0)
    x0 = np.array(x0, dtype=float)

    # Build the initial simplex (dim+1 points for a dim-D problem)
    simplex = [x0]
    for i in range(dim):
        pt = x0.copy()
        pt[i] += step
        simplex.append(pt)
    simplex = np.array(simplex)

    history = [simplex.copy()]  # store every simplex for plotting

    for it in range(max_iter):
        # 1. Order points by function value (best to worst)
        scores = np.array([f(p) for p in simplex])
        order = np.argsort(scores)
        simplex = simplex[order]
        scores = scores[order]

        best, worst, second_worst = scores[0], scores[-1], scores[-2]

        # Convergence check
        if np.std(scores) < tol:
            break

        # 2. Centroid of all points except the worst
        centroid = simplex[:-1].mean(axis=0)

        # 3. Reflection
        x_r = centroid + alpha * (centroid - simplex[-1])
        f_r = f(x_r)

        if scores[0] <= f_r < second_worst:
            simplex[-1] = x_r

        elif f_r < best:
            # 4. Expansion
            x_e = centroid + gamma * (x_r - centroid)
            f_e = f(x_e)
            simplex[-1] = x_e if f_e < f_r else x_r

        else:
            # 5. Contraction
            x_c = centroid + rho * (simplex[-1] - centroid)
            f_c = f(x_c)
            if f_c < worst:
                simplex[-1] = x_c
            else:
                # 6. Shrink toward best point
                simplex = simplex[0] + sigma * (simplex - simplex[0])

        history.append(simplex.copy())

    return simplex[np.argmin([f(p) for p in simplex])], history


# Run the optimizer, starting far from the minimum
x0 = [-4, 8]
best_point, history = nelder_mead(f, x0, step=1.5)
print(f"Optimal point found: {best_point}, f = {f(best_point):.4f}")
print(f"Converged in {len(history)} simplex steps")


# Visualize the terrain + every simplex step
x_range = np.linspace(-10, 10, 200)
y_range = np.linspace(-10, 10, 200)
X, Y = np.meshgrid(x_range, y_range)
Z = f([X, Y])

plt.figure(figsize=(9, 7))
plt.contourf(X, Y, Z, levels=50, cmap='terrain')
plt.colorbar(label='Height (m)')

n_steps = len(history)
colors = cm.cool(np.linspace(0, 1, n_steps))  # blue = early, pink = late

for i, simplex in enumerate(history):
    triangle = np.vstack([simplex, simplex[0]])  # close the triangle
    plt.plot(triangle[:, 0], triangle[:, 1],
              color=colors[i], linewidth=1.2, alpha=0.8, marker='o', markersize=3)

# Mark start and final optimum
plt.plot(*x0, 'ks', markersize=10, label='Start')
plt.plot(*best_point, 'r*', markersize=18, label='Optimum found')

plt.title("Nelder-Mead Simplex tracked")
plt.xlabel('East (x)')
plt.ylabel('North (y)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm


# Define the terrain function
def f(coords):
    x, y = coords
    return (x**2 - 2)**2 - x*y + (y**2 - 3)**2


# Custom Nelder-Mead that logs every simplex step

def nelder_mead(f, x0, step=1.0, tol=1e-6, max_iter=60,
                 alpha=1.0, gamma=2.0, rho=0.5, sigma=0.5):
    dim = len(x0)
    x0 = np.array(x0, dtype=float)

    simplex = [x0]
    for i in range(dim):
        pt = x0.copy()
        pt[i] += step
        simplex.append(pt)
    simplex = np.array(simplex)

    history = [simplex.copy()]

    for it in range(max_iter):
        scores = np.array([f(p) for p in simplex])
        order = np.argsort(scores)
        simplex = simplex[order]
        scores = scores[order]

        best, worst, second_worst = scores[0], scores[-1], scores[-2]

        if np.std(scores) < tol:
            break

        centroid = simplex[:-1].mean(axis=0)
        x_r = centroid + alpha * (centroid - simplex[-1])
        f_r = f(x_r)

        if scores[0] <= f_r < second_worst:
            simplex[-1] = x_r
        elif f_r < best:
            x_e = centroid + gamma * (x_r - centroid)
            f_e = f(x_e)
            simplex[-1] = x_e if f_e < f_r else x_r
        else:
            x_c = centroid + rho * (simplex[-1] - centroid)
            f_c = f(x_c)
            if f_c < worst:
                simplex[-1] = x_c
            else:
                simplex = simplex[0] + sigma * (simplex - simplex[0])

        history.append(simplex.copy())

    return simplex[np.argmin([f(p) for p in simplex])], history


# Run TWO searches: original start, and its mirror (-x, -y)
x0_a = [-4, 8]
x0_b = [4, -8]   # mirrored starting point -> pulls toward the OTHER minimum

best_a, history_a = nelder_mead(f, x0_a, step=1.5)
best_b, history_b = nelder_mead(f, x0_b, step=1.5)

print(f"Run A: start {x0_a} -> optimum {best_a}, f = {f(best_a):.4f}")
print(f"Run B: start {x0_b} -> optimum {best_b}, f = {f(best_b):.4f}")


# Visualize the terrain + both simplex walks
x_range = np.linspace(-10, 10, 200)
y_range = np.linspace(-10, 10, 200)
X, Y = np.meshgrid(x_range, y_range)
Z = f([X, Y])

plt.figure(figsize=(9, 7))
plt.contourf(X, Y, Z, levels=50, cmap='terrain')
plt.colorbar(label='Height (m)')

def plot_history(history, cmap):
    n_steps = len(history)
    colors = cmap(np.linspace(0, 1, n_steps))
    for i, simplex in enumerate(history):
        triangle = np.vstack([simplex, simplex[0]])
        plt.plot(triangle[:, 0], triangle[:, 1],
                  color=colors[i], linewidth=1.0, alpha=0.8, marker='o', markersize=2)

plot_history(history_a, cm.cool)
plot_history(history_b, cm.autumn)

plt.plot(*x0_a, 'ks', markersize=10, label='Start A')
plt.plot(*best_a, 'b*', markersize=18, label='Optimum A')
plt.plot(*x0_b, 'k^', markersize=10, label='Start B')
plt.plot(*best_b, 'r*', markersize=18, label='Optimum B (other spot)')

plt.title("Nelder-Mead: Two Starting Points Finding different Optima")
plt.xlabel('East (x)')
plt.ylabel('North (y)')
plt.legend()
plt.show()
```
